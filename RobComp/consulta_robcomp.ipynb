## 7. Action Server e Action Client
### Action Server

Recebe um objetivo, executa uma tarefa demorada, envia feedback durante
a execução e informa o resultado final.

### Action Client

Envia o objetivo ao servidor, acompanha o feedback e recebe o resultado.

## 11. Checklist rápido de diagnóstico
Quando um nó não funcionar, confira nesta ordem:
- [ ] Estou no workspace correto?
- [ ] Executei `colcon build`?
- [ ] Executei `source install/setup.bash` neste terminal?
- [ ] O nó aparece em `ros2 node list`?
- [ ] O tópico aparece em `ros2 topic list`?
- [ ] O nome do tópico está correto?
- [ ] O tipo da mensagem está correto?
- [ ] O QoS é compatível?
- [ ] O callback possui `self` e `msg`?
- [ ] Passei `self.callback` sem parênteses?
- [ ] O publisher realmente executa `.publish()`?
- [ ] O nó está ativo com `rclpy.spin()`?
- [ ] O simulador, teleop e meu nó estão em terminais separados?
- [ ] O terminal do teleop está com o foco do teclado?
- [ ] As variáveis foram inicializadas no `__init__`?
- [ ] A indentação está correta?
- [ ] Há alguma mensagem de erro no terminal?

### 12. Configuração do Pacote (ROS 2)
- Incluir o pacote robcomp_util (clonar do github do prof ou pegar o meu) como dependencia do pacote: declarar no ROS 2 que entregavel_X depende de robcomp_util;
- Importar, nos arquivos Python, as classes que estão dentro de robcomp_util.

**Comando no terminal:**

`ros2 pkg create --build-type ament_python entregavel_X \
    --dependencies rclpy geometry_msgs sensor_msgs nav_msgs robcomp_util`

- _Se o pacote ja foi criado anteriormente: ir no arquivo_ `package.xml` _e acrescentar_ `<depend>robcomp_util</depend>`

**Imports nos códigos (apenas das classes necessárias):**

SINTAXE GERAL: `from nome_do_pacote.nome_do_arquivo import NomeDaClasse`
- Ex.: `from robcomp_util.laser import Laser`

**Obs.:Apenas os executáveis devem ficar dentro do arquivo `setup.py`**
